Some common computer vision libraries in pytorch

torchvision -- base domain for pytorch computer vision

torchvision.datasets -- get datasets and data loading functions for computer vision

torchvision.models -- get pretrained computer vision models

torchvision.transforms -- functions for manipulating vision data to be suitable for use with an ML model (turn our image data into numbers so we can use it)

torch.utils.data.Dataset -- base dataset class for pytorch

torch.utils.data.DataLoader -- Creater a python iterable over a dataset

```python
import torch
from torch import nn

import torchvision
from torchvision import datasets
from torchvision import transforms
from torchvision.transforms import ToTensor

import matplotlib.pyplot as plt
print(torch.__version__)
print(torchvision.__version__)
```

###Getting a dataset

We will be using FashionMNIST from torchvision.datasets

```python
#setting up training data

train_data = datasets.FashionMNIST(
    root = "data", #where to download data to
    train = True, # do we want the training or testing dataset
    download = True, #do we want to downlaod it?
    transform = ToTensor(), #we are converting our images to tensor
    target_transform = None #how do we want to transform the labels
)

test_data = datasets.FashionMNIST(
    root = "data",
    train = False,
    download = True,
    transform = True,
    target_transform = False
)
```
